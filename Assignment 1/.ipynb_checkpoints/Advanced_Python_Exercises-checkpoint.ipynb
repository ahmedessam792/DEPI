## 8. OOP — Task
Create a Student class.

```python
class Student :
    def __init__(self,name,age):
        self.name = name 
        self.age = age
    def display_info(self):
        print(f"Name : {self.name} , Age : {self.age}")
```

```python
std = Student("Ahmed",22)
std.display_info()
```

## 9. Inheritance — Task
Create GraduateStudent.

```python
class GraduateStudent(Student):
    def __init__(self,name,age,degree,university) : 
        super().__init__(name,age)
        self.degree = degree
        self.university = university
    def display_info(self):
        print(f"Name : {self.name} , Age : {self.age} , University : {self.university} , Degree : {self.degree}")
```

```python
std_2 = GraduateStudent("Ahmed",22,"Master","MIT")
std_2.display_info()
```

## 10. Decorators — Task
Logger decorator.

```python
def logger(func):
    def wrapper():
        print(f"Starting function: {func.__name__}")
        func()
        print(f"Finished function: {func.__name__}")
    return wrapper
@logger
def welcome():
    print("Hello Ahmed")
welcome()
```

## 11. Generators — Task
Countdown generator.

```python
def Countdown(number):
    while number > 0:
        yield number # instead of return
        number -= 1
```

```python
for value in Countdown(5):
    print(value)
```

## 12. Lambda, Map, Filter, Reduce — Task
Work with numbers list.

```python
from functools import reduce
nums = [1,2,3,4,5,6,7]
squares = list(map(lambda num : num **2, nums))
even_ones =  list(filter(lambda num: num % 2 == 0, nums))
total = reduce(lambda first, second: first + second, nums)
print("Original numbers:", nums )
print("Squares:", squares)
print("Even numbers:", even_ones)
print("Total:", total)
```

## 13. Advanced Practice
- Max/min/avg function
- Agent class
- Generator divisible by 3

```python
def calculate_nums(nums):
    maximum = max(nums)
    minimum = min(nums)
    n = len(nums)
    average = sum(nums) / n
    return maximum , minimum , average
```

```python
nums = [1,34,23,55,6,9,23]
maximum , minimum , average = calculate_nums(nums)
print("Maximum:", maximum)
print("Minimum:", minimum)
print("Average:", average)
```

```python
class Agent:
    def __init__(self,name,role,tool,goal):
        self.name = name 
        self.role = role 
        self.tool = tool 
        self.goal = goal 
    def display_info(self):
        print(f"Name : {self.name} ,Role : {self.role} ,Tool : {self.tool} ,Goal : {self.goal} ")
```

```python
agent_1 = Agent("Pdf viewer","read PDF and summarize","Microsoft edge","summarize the pdf in some bullet ")
agent_1.display_info()
```

```python
def divisible_by_three(limit):
    for number in range(1, limit + 1):
        if number % 3 == 0:
            yield number
```

```python
for value in divisible_by_three(15):
    print(value)
```
